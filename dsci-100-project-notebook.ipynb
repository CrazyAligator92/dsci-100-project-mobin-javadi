#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
library(tidymodels)

set.seed(70085998)

# split hte data into 75% training and 25% testing
mm_split <- initial_split(womens_mm_tidy, 
                          prop = 0.75, 
                          strata = tourney_wins)

mm_train <- training(mm_split)
mm_test <- testing(mm_split)

# create the linear regression model spec
mm_spec <- linear_reg() |>
    set_engine("lm") |>
    set_mode("regression")

# creaet a recipe with the three predictors
mm_recipe <- recipe(tourney_wins ~ seed + reg_wins + reg_losses,
                    data = mm_train)

# create the workflow and fit the model using training data
mm_fit <- workflow() |>
    add_recipe(mm_recipe) |>
    add_model(mm_spec) |>
    fit(data = mm_train)

mm_fit

# make predictions on the testing data and evaluate the model
mm_test_results <- mm_fit |>
    predict(mm_test) |>
    bind_cols(mm_test) |>
    metrics(truth = tourney_wins, estimate = .pred)

mm_test_results
```
